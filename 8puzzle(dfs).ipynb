```python
# Goal state
goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

# Function to generate all possible next states
def neighbors(state):

    result = []

    blank = state.index(0)

    row = blank // 3
    col = blank % 3

    moves = [(-1,0), (1,0), (0,-1), (0,1)]

    for dr, dc in moves:

        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:

            new_blank = new_row * 3 + new_col

            temp = list(state)

            temp[blank], temp[new_blank] = temp[new_blank], temp[blank]

            result.append(tuple(temp))

    return result


# Depth First Search
def dfs(start):

    stack = [(start, [])]

    visited = {start}

    while stack:

        state, path = stack.pop()

        if state == goal:
            return path + [state]

        for nxt in neighbors(state):

            if nxt not in visited:

                visited.add(nxt)

                stack.append((nxt, path + [state]))

    return None


# Function to print puzzle
def print_state(state):

    for i in range(0,9,3):
        print(state[i:i+3])

    print()


# Main Program
start = (1,2,3,
         4,0,6,
         7,5,8)

print("Initial State")
print_state(start)

print("DFS Solution")

answer = dfs(start)

for s in answer:
    print_state(s)
```
